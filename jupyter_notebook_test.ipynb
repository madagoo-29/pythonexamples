Strings contain text, created using single quote marks:

```python
string = 'Hi'
print(string)
```

The function len tells us how many characters are in the string:

```python
len(string)
```

Strings can be change to integer values using the int function, but only if the string contains just numbers (otherwise an error will occur). Conversely, integers can be converted to strings using the str function, e.g.:

```python
str(10)
```

Specific characters in a string can be extracted using indexing, similarly to R. The first character is extracted with 0, the second charcacter by 1, and so on:

```python
print(string[0])
print(string[1])
```

The operator + between two integers will sum them together, and between two strings will concatenate them together, e.g.:

```python
string + ' there'
```

We can use the operator 'in' to determine whether a character exists in a string (exact match), which will return True if the character is in the string, False if the character is not in the string, and an error otherwise:

```python
'H' in string # returns True if H (exact match) is in string
```

```python
'z' in string # returns False if z is not in string
```

Character tests can be performed. These include .isupper, .islower, .isalpha (does the string contain only letters), .isdigit, .isspace, .startswith, and .endswith. 

We can find where a character exists in a string using .find (again, the first character in a string has index 0):

```python
string.find('H') # find H (exact match) in string
```

We can trim whitespace using .strip and replace characters using .replace.

Since strings are immutable, if we want to change a string we must redefine it, e.g.:

```python
print(string)
string = string.lower() # make string lower case
print(string)
```

Backslashes are commonplace in regular expression to escape special characters (in other words, turn off the special meaning) or introduce special meaning to normal characters. For example, the expression \n produces a new line. To insert a backslash, two backslashes are required, so \\.

Tuples are objects that are again immutable, but can contain more than one item, being either strings or numbers. They are introduced using round brackets: 

```python
tup = (1, 'me', 4)
print(tup)
```

Indexing a tuple means one can access any item (whereas for a string, we were accessing individual characters):

```python
print(tup[2])
print(tup[1])
print(tup[1][1])
```

From the final line of code above, we can index the string that has been accessed from the tuple just by writing a second pair of square brackets.

A useful functionality of tuples is that multiple variables can be defined in one go:

```python
(a, b) = (20, 30)
print(a)
print(b)
```

Creating a list (an object with multiple variables, a.k.a. elements). Lists are mutable, i.e., can be edited after creation, e.g. with .append:

```python
list = [2,10,2]
print(list)
list.append(4)
print(list)
```

Lists can contain data of any type, so can include strings, numbers, as well as other lists.

Deleting the list using del:

```python
del list
```

Empty lists can be created:

```python
list = []
```

Then the append function can be used to add elements to the list:

```python
list.append("Bee")
list.append("Grin")
list.append(14)
list.append([52, 10])
list.append(False)
```

The above code will create a list with five elements, the first two elements being strings, the third element being an integer, the fourth element being another list (which contains two integers), and the final element being the logical statement False.

```python
print(list)
```

Again, as lists are mutable, we can change any element:

```python
list[0] = "Cheese"
print(list)
```

Another useful function is .pop, which removes the final element in the list. If an index position is supplied, then the element in that position is printed and removed from the list.

```python
list.pop()
print(list)
list.pop(0)
```

Similarly, the function .remove will remove the first element from the list that is equal to the specified argument:

```python
list2 = [2, 10, 10, 2]
print(list2)
list2.remove(2)
print(list2)
list2.remove(2)
print(list2)
```

The above demonstrates that it is just the first element that occurs which is equal to the argument which is removed.

The .extend function works similarly to the .append function but instead allows you to add more than one element to a list at a time:

```python
list3 = ['hi', 'there']
list3.extend(['how', 'are', 'you'])
list3
```

Note that .extend will only work if you are wanting to add a list to a list; if you want to add just a single element to a list, you have to use .append.

The + operator works in the same way, except the original list won't be changed. If the aim is to update the original list, then += can be used:

```python
list4 = [34, 10, 'pie']
list5 = ['maybe', False]
print(list4 + list5)
print(list4)
list4 += list5
print(list4)
```

Other useful functions:
- .index, allowing you to find where in the list some element sits
- .insert, allowing you to insert an element into a list at a given index (the elements at and after the index are shifted right)
- sorted(), which sorts elements in the list from smallest to largest or into alphabetical order (only works when the list contains only integers or contains only strings)
- min(), which returns the smallest integer or which string comes first alphabetically
- max(), which returns the largest integer or which string comes last alphabetically
- sum(), which adds together all integers in list (list cannot contain strings)

For-looping is particularly useful with lists, whether it be printing out each element of a list in turn, or do some maniuplation of each element of a list of data points. For-loop notation is slightly different to R:

```python
list6 = [1, 3, 5, 6, 9]
for i in range(len(list6)): # for some variable i taking the value 0, then 1, ..., up to length of list6
    print(list6[i]) # print each element
```

A common method for creating a list with data is to create an empty list as done previously, then write a for-loop which iteratively appends (so using .append) data points to the list.
