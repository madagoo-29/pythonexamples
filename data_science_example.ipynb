This notebook considers the year-on-year predictive power of various variables with respect to goalscoring in the Barclays Women's Super League, the top professional football league in women's football in England. The purpose of this document is not to demonstrate novel statistical analysis of football, rather to demonstrate an example of 'data science' being applied to real-world data.

The document assumes the reader has a basic understanding of football, including knowledge of counting statistics (shots, goals, minutes played) that are collected during football matches. These statistics are extracted from multiple websites (Fotmob, fbref) for the 2024/2025 and 2025/2026 football seasons, with the aim of investigating whether we can reliably predict goal scoring season on season.
